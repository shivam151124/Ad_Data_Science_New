## 1. Import the dataset

```python
import pandas as pd

df = pd.read_csv("trends.csv")
print("Dataset loaded successfully.")
display(df.head())
```

## 2. Identify and handle missing values

```python
print("Missing values:")
print(df.isnull().sum())

# Handle missing values
df = df.dropna()

print("Missing values handled.")
```

## 3. Remove duplicate records

```python
print("Duplicates before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after:", df.duplicated().sum())
```

## 4. Correct inconsistent entries

```python
# Correct inconsistent text entries

for col in df.select_dtypes(include=["object", "string"]).columns:
    df[col] = df[col].astype("string").str.strip().str.lower()

print("Inconsistent entries corrected.")
```

## 5. Assign appropriate data types

```python
# Convert numeric columns
df["year"] = pd.to_numeric(df["year"])
df["rank"] = pd.to_numeric(df["rank"])

print(df.dtypes)
```

## 6. Rename columns where necessary

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

print("Column names:")
print(df.columns.tolist())
```

## 7. Check and correct formatting issues

```python


for col in df.select_dtypes(include=["object", "string"]).columns:
    df[col] = df[col].astype("string").str.strip()

print("Formatting checked and corrected.")
```

## 8. Verify the final cleaned dataset

```python
print("Final shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate records:", df.duplicated().sum())
print("\nData types:")
print(df.dtypes)

display(df.head())
```

## Save the cleaned dataset

```python
df.to_csv("trends_cleaned.csv", index=False)
print("Cleaned dataset saved as trends_cleaned.csv")
```
