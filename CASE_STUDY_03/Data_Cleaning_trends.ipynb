# Data Importing and Basic Data Cleaning

**Dataset:** `trends.csv`

This notebook performs the required 8 data cleaning tasks:
1. Import the dataset
2. Identify and handle missing values
3. Remove duplicate records
4. Correct inconsistent entries
5. Assign appropriate data types
6. Rename columns where necessary
7. Check and correct formatting issues
8. Verify the final cleaned dataset

> Keep `trends.csv` in the same folder as this notebook before running it.

```python
import pandas as pd

# 1. Import the dataset
df = pd.read_csv("trends.csv")

print("Dataset loaded successfully.")
print("Shape:", df.shape)
display(df.head())
```

## 2. Identify and Handle Missing Values

```python
print("Missing values before cleaning:")
print(df.isnull().sum())
print("Total missing values:", df.isnull().sum().sum())

# Handle missing values safely
# Numeric columns -> median
for col in df.select_dtypes(include="number").columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

# Text columns -> mode
for col in df.select_dtypes(include=["object", "string"]).columns:
    if df[col].isnull().any():
        mode_value = df[col].mode()
        if not mode_value.empty:
            df[col] = df[col].fillna(mode_value.iloc[0])

print("\nMissing values after cleaning:")
print(df.isnull().sum())
print("Total missing values:", df.isnull().sum().sum())
```

## 3. Remove Duplicate Records

```python
duplicate_count = df.duplicated().sum()
print("Duplicate rows before removal:", duplicate_count)

df = df.drop_duplicates().reset_index(drop=True)

print("Duplicate rows after removal:", df.duplicated().sum())
print("New shape:", df.shape)
```

## 4. Correct Inconsistent Entries

```python
# Remove leading/trailing spaces from text values
text_cols = df.select_dtypes(include=["object", "string"]).columns

for col in text_cols:
    df[col] = df[col].astype("string").str.strip()

print("Text entries were checked and unnecessary spaces were removed.")

# Show sample unique values for categorical columns
for col in ["location", "category"]:
    if col in df.columns:
        print(f"\nSample values in {col}:")
        print(df[col].dropna().unique()[:10])
```

## 5. Assign Appropriate Data Types

```python
# Convert numeric columns appropriately when they exist
if "year" in df.columns:
    df["year"] = pd.to_numeric(df["year"], errors="coerce")

if "rank" in df.columns:
    df["rank"] = pd.to_numeric(df["rank"], errors="coerce")

# Convert text columns to string
for col in ["location", "category", "query"]:
    if col in df.columns:
        df[col] = df[col].astype("string")

print("Data types after correction:")
print(df.dtypes)
```

## 6. Rename Columns Where Necessary

```python
# Standardize column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
)

print("Final column names:")
print(df.columns.tolist())
```

## 7. Check and Correct Formatting Issues

```python
# Remove extra spaces again after column standardization
for col in df.select_dtypes(include=["object", "string"]).columns:
    df[col] = df[col].astype("string").str.strip()

# Check for blank text values
blank_counts = {}
for col in df.select_dtypes(include=["object", "string"]).columns:
    blank_counts[col] = (df[col].fillna("").str.strip() == "").sum()

print("Blank text values:")
print(blank_counts)

print("\nFormatting check completed.")
```

## 8. Verify the Final Cleaned Dataset

```python
print("=" * 50)
print("FINAL DATASET VERIFICATION")
print("=" * 50)

print("\nFinal Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Records:")
display(df.head())

print("\nLast 5 Records:")
display(df.tail())
```

## Save the Cleaned Dataset

```python
# Save cleaned dataset
output_file = "trends_cleaned.csv"
df.to_csv(output_file, index=False)

print(f"Cleaned dataset saved successfully as: {output_file}")
```

## Brief Summary

- The `trends.csv` dataset was imported using Pandas.
- Missing values were identified and handled where required.
- Duplicate records were identified and removed.
- Text entries were checked for unnecessary spaces and standardized.
- Appropriate data types were assigned to the columns.
- Column names were standardized using lowercase and underscores.
- Formatting issues were checked.
- The final cleaned dataset was verified.
- The cleaned dataset was saved as `trends_cleaned.csv`.
