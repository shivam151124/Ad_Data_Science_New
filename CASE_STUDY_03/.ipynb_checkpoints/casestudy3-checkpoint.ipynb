# Initial Data Analysis – trends.csv

This notebook performs an initial analysis of `trends.csv` using Python Pandas. It checks the dataset structure, data types, missing values, duplicates, unique values, and basic statistical measures.

```python
import pandas as pd

df = pd.read_csv("trends.csv")
print("Dataset loaded successfully.")
```

**Explanation:** The CSV file is loaded into a Pandas DataFrame named `df`.

```python
print(df.head())
```

**Explanation:** `head()` displays the first five rows. It is useful for checking the initial records and confirming that the dataset was loaded correctly.

```python
print(df.tail())
```

**Explanation:** `tail()` displays the last five rows. It helps inspect the end of the dataset for unexpected or missing values.

```python
print("Shape:", df.shape)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

**Explanation:** `shape` returns the number of rows and columns. This dataset has 26,955 rows and 5 columns.

```python
print(df.columns.tolist())
```

**Explanation:** The `columns` attribute lists all variables in the dataset.

```python
print(df.dtypes)
```

**Explanation:** `dtypes` shows the data type of each column. These types determine how the variables can be analyzed and whether conversion may be needed.

```python
df.info()
```

**Explanation:** `info()` summarizes the DataFrame, including row count, non-null counts, data types, and memory usage. Comparing non-null counts with the total rows helps identify missing data.

```python
print(df.describe())
```

**Explanation:** `describe()` gives numerical summary statistics such as count, mean, standard deviation, minimum, quartiles, and maximum for numeric columns.

```python
print(df.describe(include="all").T)
```

**Explanation:** `describe(include="all")` extends the summary to categorical/text columns as well. It can show counts, unique values, most frequent values, and frequencies for non-numeric columns.

```python
missing = df.isnull().sum()
print(missing)
print("Total missing values:", missing.sum())
```

**Explanation:** `isnull().sum()` counts missing values in every column. Columns with non-zero counts require further cleaning or an appropriate missing-value treatment.

```python
duplicates = df.duplicated().sum()
print("Duplicate rows:", duplicates)
```

**Explanation:** `duplicated().sum()` counts complete duplicate records. Duplicate rows may need to be removed if they are accidental repetitions.

```python
print(df.nunique())
```

**Explanation:** `nunique()` reports the number of distinct values in each column. It helps identify categorical fields and variables with very low or very high diversity.

```python
for col in df.columns:
    print(f"\n{col}: {df[col].nunique()} unique values")
    print(df[col].dropna().unique()[:10])
```

**Explanation:** This displays the first 10 distinct values from each column as a compact inspection. It avoids printing thousands of search terms while still showing the type of values present.

```python
numeric_cols = df.select_dtypes(include="number").columns
print("Numerical columns:", list(numeric_cols))
print("\nMean:")
print(df[numeric_cols].mean())
print("\nMedian:")
print(df[numeric_cols].median())
print("\nMinimum:")
print(df[numeric_cols].min())
print("\nMaximum:")
print(df[numeric_cols].max())
print("\nStandard deviation:")
print(df[numeric_cols].std())
```

**Explanation:** These calculations provide the requested mean, median, minimum, maximum, and standard deviation for every numerical column.

```python
print("Missing values by column:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
print("\nData types:")
print(df.dtypes)
```

**Explanation:** This final data-quality check summarizes the main cleaning issues found during the initial analysis.

## Overall Interpretation

The initial analysis provides a structural and quality overview of `trends.csv`. The dataset contains location, year, category, rank, and query information. The numerical variables can be summarized statistically, while the categorical variables can be examined through their unique-value counts and frequency information. Any missing values, duplicate rows, unusual ranges, or inappropriate data types identified in the outputs should be considered during the next data-cleaning stage.
