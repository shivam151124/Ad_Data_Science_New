```python
import pandas as pd
df = pd.read_csv("trends.csv")
print(df.head())
print(df.tail())
print(df.info())
print(df.describe())
print(df.shape)
print(df["query"])
df
```

```python
columnname='rank'
Q1 = df[columnname].quantile(0.25)
Q3 = df[columnname].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR



outliers = df[(df[columnname] < lower) | (df[columnname] > upper)]
clean_iqr = df[(df[columnname] >= lower) & (df[columnname] <= upper)]
print(f"Cleaned DataFrame rows: {len(clean_iqr)}")

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower)
print("Upper Limit:", upper)

print("Outliers:")
print(outliers)

print(outliers[columnname])
print("Number of outliers:", len(outliers))
```

```python
columnname = 'rank'

mean = df[columnname].mean()
std = df[columnname].std()

lower = mean - 3 * std
upper = mean + 3 * std

outliers = df[(df[columnname] < lower) | (df[columnname] > upper)]
clean_mean = df[(df[columnname] >= lower) & (df[columnname] <= upper)]

print(f"Cleaned DataFrame rows: {len(clean_mean)}")

print("Mean:", mean)
print("Standard Deviation:", std)
print("Lower Limit:", lower)
print("Upper Limit:", upper)

print("Outliers:")
print(outliers)

print("Outlier values:")
print(outliers[columnname])

print("Number of outliers:", len(outliers))
```

```python
columnname = 'rank'


data = sorted(df[columnname].dropna())


n = len(data)

if n % 2 == 0:
    Q2 = (data[n//2 - 1] + data[n//2]) / 2
    left = data[:n//2]
    right = data[n//2:]
else:
    Q2 = data[n//2]
    left = data[:n//2]
    right = data[n//2 + 1:]



def median(data):
    n = len(data)

    if n % 2 == 0:
        return (data[n//2 - 1] + data[n//2]) / 2
    else:
        return data[n//2]



Q1 = median(left)
Q3 = median(right)


IQR = Q3 - Q1


lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR


outliers = df[
    (df[columnname] < lower) |
    (df[columnname] > upper)
]


clean_iqr = df[
    (df[columnname] >= lower) &
    (df[columnname] <= upper)
]


print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Limit:", lower)
print("Upper Limit:", upper)

print("Outliers:")
print(outliers)

print("Number of outliers:", len(outliers))

print("Cleaned DataFrame rows:", len(clean_iqr))
```
